# Product Information Extraction and JSON Creation


Welcome to this practical exercise on extracting product information using regular expressions in Python! In this assignment, you will work with a list of products and their details from an online store. Your goal is to parse this information, compare product features, and create a structured JSON format.

## Task Overview
1. **Read product data from a text file.**
2. **Use regular expressions to extract product details.**
3. **Create a JSON representation of the products.**
4. **Output the results.**


## Step 1: Read the File
In this step, you will write Python code to read the contents of the `products.txt` file.




```python
import re
import json

# Replace the path below with the actual path to your file
file_path = r'C:\Desktop\products.txt'

# Read the content from the text file
with open(file_path, 'r') as file:
    text = file.read()
```

# Step 2: Extract Product Details
Now, you will define regular expressions to extract product names, prices, URLs, and features from the text.

```python
product_name_pattern = r'^.*? -'
description_pattern = r'-The (.*?), comes with|featuring .*?\.'
price_pattern = r'EGP\s[\d,]+'
link_pattern = r'https://www\.jumia\.com\.eg/[^\s]+'
```

# Step 3: Find All Matches
Use the defined regex patterns to find matches for product names, prices, URLs, and features.#

```python
product_names = re.findall(product_name_pattern, text, re.MULTILINE)
descriptions = re.findall(description_pattern, text)
prices = re.findall(price_pattern, text)
links = re.findall(link_pattern, text)
```

# Step 4: Create a JSON Representation
Now, let’s organize the extracted information into a structured JSON format.

```python
import json

# Function to clean and convert price strings to float
def clean_price(price_str):
    # Remove currency symbols and commas, then convert to float
    cleaned_price = price_str.replace("EGP", "").replace(",", "").strip()
    return float(cleaned_price)

# Step 4: Create a JSON Representation
products = []
for i in range(len(product_names)):
    try:
        product = {
            "name": product_names[i].strip(" -"),  
            "description": descriptions[i].strip() if i < len(descriptions) else "No description available", 
            "price": clean_price(prices[i]),  
            "link": links[i]
        }
        products.append(product)
    except ValueError as e:
        print(f"Error processing product at index {i}: {e}")

# Save the JSON output to a file
output_file_path = r'C:\Desktop\products.json'  
with open(output_file_path, 'w') as json_file:
    json.dump(products, json_file, indent=4)
```
