# Exercise 2: Find Next Square
Create a method that finds the next integral perfect square after the one passed as a parameter. Recall that an integral perfect square is an integer n such that sqrt(n) is also an integer.

If a negative prameter is passed then return -1

```
121 --> 144
625 --> 676
-76 --> -1
```



````python
import math

def FindNextSquare(num):
  # Your code here
    if num < 0 or math.sqrt(num) % 1 != 0: 
            return -1
    next_square = int(math.sqrt(num) + 1) ** 2
    return next_square

# Examples
print(FindNextSquare(121))  # Output: 144
print(FindNextSquare(625))  # Output: 676
print(FindNextSquare(-76))  # Output: -1
````

# Exercise 3: Find the unique number
There is an array with some numbers. All numbers are equal except for one. Try to find it!



```
find_uniq([ 1, 1, 1, 2, 1, 1 ]) == 2
find_uniq([ 0, 0, 0.55, 0, 0 ]) == 0.55
```



````python
def find_uniq(List_Num):
  # Your code here
    for num in set(List_Num):
        if List_Num.count(num) == 1:
            return num

# Examples
print(find_uniq([1, 1, 1, 2, 1, 1]))  # Output: 2
print(find_uniq([0, 0, 0.55, 0, 0]))  # Output: 0.55
````

# Exercise 4: List Filtering
Create a function that takes a list of non-negative integers and strings and returns a new list with the strings filtered out.

```
filter_list([1,2,'a','b']) == [1,2]
filter_list([1,'a','b',0,15]) == [1,0,15]
filter_list([1,2,'aasf','1','123',123]) == [1,2,123]
```



````python
def filter_list(List):
  # Your code here
  return [item for item in List if isinstance(item, int)]

# Examples
print(filter_list([1, 2, 'a', 'b']))  # Output: [1, 2]
print(filter_list([1, 'a', 'b', 0, 15]))  # Output: [1, 0, 15]
print(filter_list([1, 2, 'aasf', '1', '123', 123]))  # Output: [1, 2, 123]
````

# Exercise 5: Vowel Count
Return the number (count) of vowels in the given string.

We will consider a, e, i, o, u as vowels.

The input string will only consist of lower case letters and/or spaces.



````python
# Your code here
def count_vowels(string):
    vowels = "aeiou"
    return sum(1 for char in string if char in vowels)

# Examples
print(count_vowels("aeiooolan"))  # aeiooolan ==> 7
````

# Exercise 7: Enumerate
For the following list:
```
List = [1, "Debi", "Hello", [7,2,(5,6,7)], (17.3,76)]
```
Loop over each element and print the index of the element and its contents.
Example:
```
Index 0: 1
Index 1: "Debi"
...
```
Tip: Search for enumerate function

````python
lst = [1, "Debi", "Hello", [7, 2, (5, 6, 7)], (17.3, 76)]

for index, element in enumerate(lst):
    print(f"Index {index}: {element}")
````
