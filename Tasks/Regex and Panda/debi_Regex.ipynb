# RegEx
A RegEx, or Regular Expression, is a sequence of characters that forms a search pattern.

RegEx can be used to check if a string contains the specified search pattern.

REGEX symbols<br>
- |   =>  Or
- []  =>  set ex: [a-d] == a|b|c|d
- .   =>  any character
- {}  =>  Can be used as a counter for the . symbol
- ^   =>  start of the line
- \$   =>  end of the line
- \*   =>  from 0 to infinite characters
- \+   =>  from 1 to infinite characters
- ()  =>  to group sub-expressions, e.g., (abc)*
- \   =>  cancel next character
- \A  =>  start of input
- \Z  =>  end of input
- \b  =>  either at the start or the end of a whole word
- \B  =>  in the middle not in start or middle of a whole word
- \w  =>  [0-9a-zA-Z_]
- \W  =>  special characters )(*&^%$#@!
- \d  =>  decimals [0-9]
- \D  =>  no decimals
- \s  =>  space, space characters are [ \n\r\t\f]
- \S  =>  no space

# Imports
Python has a built-in package called re, which can be used to work with Regular Expressions.


```python
import re # We will need this to use regex
```

# Basic search

```python
"""
re.search(pattern, text) >> searches for first occurance of pattern in text
"""
txt = "The rain in Spain"
x= re.search("ai", txt)
print(x)
```

```python
"""
re.findall(pattern, text) >> returns array of all occurances of pattern in text
"""
txt = "The rain in Spain"
x= re.findall("ai", txt)
print(x)
```

# RegEx symbols

```python
# OR Operator (|): E.g., the regex four|4 accepts strings "four" or "4".

txt = "four cars and 4 birds"
x= re.findall("four|4|and", txt)
print(x)
```

```python
# [...]: Accept ANY ONE of the character within the square bracket,
# e.g., [aeiou] matches "a", "e", "i", "o" or "u".

txt = "four cars and 4 birds"
x= re.findall("[aeiou]", txt)
print(x)

# e.g., [a-d] matches "a", "b", "c", "d"
txt = "four cars and 4 birds"
x= re.findall("[abcd]", txt)
print(x)
```

```python
'''
+: one or more (1+)
*: zero or more (0+)
?: zero or one (optional)
'''
txt = "playing football"
x= re.findall("[a-z]+", txt)
print(x)
```

```python
'''
. (dot): ANY ONE character except newline.
\w: [0-9a-zA-Z_]
\W: special characters )(*&^%$#@!
\d: decimals [0-9]
\D: no decimals
\s: space, space characters are [ \n\r\t\f]
\S: no space
'''

txt = "seven cars and birds]"
x= re.findall("\w+", txt)
print(x)
```

```python
'''
^ : start of the line
$ : end of the line
\A: start of input
\Z: end of input
\b: either at the start or the end of a whole word
\B: in the middle not in start or end of a whole word
'''

txt = "A categorical cat"
x= re.findall(r"cat\B", txt)
print(x)
```

```python
# Function for word tokenization
def tokenize_text(text):
    # Regex to match words, including handling apostrophes within words (e.g., "It's")
    tokens = re.findall(r"\b\w+(?:'\w+)?\b", text)
    return tokens
```

# Bonus Explanation: Lookaheads and Lookbehinds

## Lookahead Assertions:
- `(?=...)`: Positive lookahead matches if the pattern specified inside `(?=...)` follows the current match.
- `(?!...)`: Negative lookahead matches if the pattern specified inside `(?!...)` does **not** follow the current match.

## Lookbehind Assertions:
- `(?<=...)`: Positive lookbehind matches if the pattern specified inside `(?<=...)` precedes the current match.
- `(?<!...)`: Negative lookbehind matches if the pattern specified inside `(?<!...)` does **not** precede the current match.

### How They Work:
- Lookaheads and lookbehinds are non-consuming assertions, meaning they match conditions at a specific position without including those conditions in the result.

```python
'''
Positive Lookahead (?=...):
Matches if the specified pattern follows the match.
'''

txt = "He is playing football and running in the park"


x = re.findall(r"\w+(?=ing)", txt)
print(x)



# Find words that are followed by "ing"
# x = re.findall(r"\b[a-z]+(?=ing\b)", txt)
print(x)  # Prints words like 'play', 'runn' without including 'ing'

# Quick Question: how to return the whole word including "ing"?
```

```python
'''
Negative Lookahead (?!...):
Matches if the specified pattern does not follow the match.
'''



txt = "He is playing football and running in the park"
x = re.findall(r"[a-z]+(?!ing)", txt)
print(x)


# Find words that do not end with "ing"
x = re.findall(r"\b[a-z]+\b(?!ing\b)", txt)
# print(x)  # Prints words that do not end with 'ing'
```

```python
'''
Positive Lookbehind (?<=...):
Matches if the specified pattern precedes the match.
'''

txt = "cat dog category fat running."
# Find words that starts  by 'cat'
x = re.findall(r"(?=cat)[a-z]+", txt)
print(x)  # Prints words starting by cat

# Quick Question: how to print words starting by cat but remove the letters "cat" from it?
```

```python
'''
Negative Lookbehind (?<!...):
Matches if the specified pattern does not precede the match.
'''
txt = "The cat was playing, but the dog was not."
# Find 'was' only if it is NOT preceded by 'The'
x = re.findall(r"(?<!cat )w", txt)
print(x)  # Prints 'was' if 'The' is not right before it
```
